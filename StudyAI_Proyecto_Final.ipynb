### **StudyAI — Asistente Inteligente para la Generación de Planes de Estudio**

**StudyAI** es una Proof of Concept (POC) multimodal diseñada para transformar la desestructuración y sobrecarga cognitiva de los estudiantes en cronogramas de estudio hiperpersonalizados, pedagógicamente rigurosos y visualmente sintetizados, combinando modelos de lenguaje de última generación (*OpenAI GPT*) con síntesis visual generativa (*Texto-a-Imagen*).


## 3. Resumen

El presente proyecto aborda una de las dificultades más universales y persistentes en el ámbito académico: la incapacidad de transformar un programa de examen o lista de temas en un plan de acción viable, realista y ajustado a la disponibilidad temporal de cada estudiante. Con frecuencia, los planes tradicionales fracasan debido a la subestimación del tiempo necesario, la ausencia de momentos sistemáticos de consolidación y la falta de jerarquización de contenidos.

Para dar respuesta a esta problemática, se desarrolló **StudyAI**, una Proof of Concept (POC) fundamentada en Inteligencia Artificial Generativa y en la disciplina del **Fast Prompting**. El sistema toma como variables de entrada el perfil particular del estudiante (materia, temas, días disponibles, horas diarias, nivel de competencia, fecha límite, metas académicas y preferencias metodológicas) y, mediante un diseño de prompts iterativo y metodológicamente fundado (V1 Básico $\rightarrow$ V2 Estructurado $\rightarrow$ V3 Optimizado), produce un plan pedagógico integral organizado en 9 secciones obligatorias que incorpora técnicas activas de aprendizaje como la *Repetición Espaciada (Spaced Repetition)* y la *Evocación Activa (Active Recall)*.

De forma complementaria, el proyecto incorpora un modelo texto-a-imagen para sintetizar el plan en una infografía visual en formato de calendario semanal, aprovechando la teoría de la doble codificación de Paivio para optimizar la retención y reducir la fricción cognitiva. La solución se implementó en Python con gestión segura de credenciales y un modo de simulación académica determinista que garantiza la reproducibilidad total e inmediata para cualquier evaluador.

## 5. Problemática

La planificación del estudio universitario y profesional enfrenta barreras cognitivas y metodológicas bien documentadas:

1. **Parálisis por Análisis y Sobrecarga Cognitiva:** Al enfrentarse a programas extensos con múltiples dependencias conceptuales, el estudiante no sabe por dónde comenzar ni cómo dosificar los temas.
2. **La Falacia de la Planificación:** Tendencia sistemática a subestimar el tiempo requerido para comprender conceptos abstractos o resolver ejercicios prácticos, lo que lleva a jornadas maratónicas e ineficaces en los días previos a la evaluación.
3. **Lectura Pasiva e Ilusión de Competencia:** La mayoría de los estudiantes dedican su tiempo a releer apuntes o subrayar textos de manera pasiva, métodos que la ciencia cognitiva ha demostrado altamente ineficientes en comparación con la evocación activa.
4. **Falta de Repaso Sistemático:** Al no planificar repasos acumulativos, los conocimientos adquiridos en los primeros días se degradan rápidamente por el efecto de la curva del olvido (Ebbinghaus).
5. **Planes Rígidos e Inviables:** Los cronogramas habituales suelen ser listas planas que no consideran pausas cognitivas, niveles de partida ni preferencias personales, lo que provoca frustración y abandono temprano.

## 11. Metodología

La investigación y desarrollo se basó en el ciclo iterativo de **Fast Prompting**, consistente en cuatro etapas recurrentes:

1. **Definición de Línea Base (Baseline):** Formulación de un prompt V1 en lenguaje natural informal, sin restricciones explícitas ni rol, para observar el comportamiento por defecto del modelo.
2. **Diagnóstico Crítico de Brechas:** Análisis sistemático de las omisiones de V1 (falta de granularidad temporal, ausencia de pausas y repasos, formato no estandarizado).
3. **Inyección Estructurada de Técnicas:** Incorporación progresiva en V2 y V3 de delimitadores, directrices de rol, anclas contextuales, micro-ejemplos *few-shot* y restricciones numéricas.
4. **Validación Cualitativa y Ajuste:** Comparación transversal de los resultados mediante una matriz de evaluación que mide la claridad, consistencia y aplicabilidad práctica del plan.

## 13. Técnicas de Fast Prompting

En StudyAI, ninguna técnica se añade con fines meramente ornamentales. Cada una cumple una función arquitectónica y pedagógica insustituible:

| Técnica | Función en StudyAI | Problema que Resuelve |
|:---|:---|:---|
| **Role Prompting** | Asigna la identidad de *Asesor Académico de Élite y Especialista en Ciencia del Aprendizaje*. | Evita respuestas genéricas o coloquiales; calibra el tono al ámbito universitario. |
| **Context Prompting** | Inyecta las 8 variables del estudiante (materia, temas, tiempo, nivel, etc.). | Elimina la ambigüedad situacional y ancla el razonamiento a la realidad del alumno. |
| **Instruction Prompting** | Imparte órdenes directas y secuenciales de clasificación y distribución temporal. | Evita divagaciones y asegura que el modelo se enfoque en planificar y no solo en describir. |
| **Few-Shot Prompting** | Muestra un micro-patrón del desglose minuto a minuto de una jornada de 2 horas. | Fuerza la granularidad temporal sin ambigüedades y estandariza la sintaxis de salida. |
| **Structured Prompting** | Emplea encabezados Markdown, separadores `===` y tablas comparativas. | Optimiza la atención del LLM y permite al estudiante navegar rápidamente el plan. |
| **Restricciones Explícitas (*Grounding*)** | Prohíbe inventar materias/temas ajenos y exige que la suma horaria sea exactamente la indicada. | Erradica la alucinación de contenidos y evita la sobrecarga de agendas inviables. |
| **Descomposición de Tareas (*Task Decomposition*)** | Divide la generación del plan en las 9 fases obligatorias requeridas. | Permite al modelo razonar paso a paso (*Chain of Thought* implícito) sobre la pedagogía. |
| **Formato de Salida Obligatorio** | Especifica la estructura de títulos y tablas esperadas. | Garantiza la consistencia estructural requerida para integración automática. |
| **Iteración y Refinamiento** | Evolución deliberada y documentada a través de V1, V2 y V3. | Valida experimentalmente el impacto de cada capa de refinamiento sobre el resultado final. |

### Funciones Auxiliares: Constructor Dinámico y Ejecutor de Prompts


````python
def construir_prompt(ruta_plantilla, variables):
    """
    Carga una plantilla de texto desde prompts/ e inyecta dinámicamente
    las variables del perfil y las métricas temporales derivadas.
    """
    with open(ruta_plantilla, 'r', encoding='utf-8-sig') as f:
        plantilla = f.read()
        
    minutos_dia = variables["horas_por_dia"] * 60
    total_horas = variables["dias_disponibles"] * variables["horas_por_dia"]
    
    mapeo = {
        "{MATERIA}": variables["materia"],
        "{TEMAS}": variables["temas"],
        "{DIAS_DISPONIBLES}": str(variables["dias_disponibles"]),
        "{HORAS_POR_DIA}": str(variables["horas_por_dia"]),
        "{MINUTOS_TOTALES_DIA}": str(minutos_dia),
        "{TOTAL_HORAS}": str(total_horas),
        "{NIVEL}": variables["nivel"],
        "{FECHA_EXAMEN}": variables["fecha_examen"],
        "{OBJETIVO}": variables["objetivo"],
        "{PREFERENCIAS}": variables["preferencias"]
    }
    
    texto_final = plantilla
    for etiqueta, valor in mapeo.items():
        texto_final = texto_final.replace(etiqueta, valor)
    return texto_final

def simular_respuesta_academica(version):
    """
    Retorna el resultado real de laboratorio registrado para cada versión
    del prompt cuando se opera en modo offline sin API key.
    """
    if version == "V1":
        return """¡Hola! Claro, para estudiar Programación en 7 días con 2 horas por día, te recomiendo este plan:

Día 1: Variables y tipos de datos.
Día 2: Condicionales (if/else).
Día 3: Bucles (for y while).
Día 4: Funciones y parámetros.
Día 5: Arrays y métodos de listas.
Día 6: Objetos y propiedades.
Día 7: Programación orientada a objetos y repaso final.

Consejos generales:
- Lee la teoría con atención.
- Haz ejercicios en tu computadora para practicar.
- Repasa los temas que te resulten más difíciles antes de la fecha del examen.
- No te olvides de descansar bien. ¡Mucho éxito en tu examen!"""

    elif version == "V2":
        return """# Plan de Estudio Estructurado: Programación

Como asistente académico universitario especializado en planificación estratégica de estudio, he diseñado el siguiente cronograma de 7 días ajustado a tu nivel intermedio y respetando estrictamente el límite de 2 horas diarias (14 horas en total).

## Resumen del Cronograma
- **Materia:** Programación
- **Duración Total:** 7 días consecutivos
- **Dedicación Diaria:** 2 horas por jornada
- **Enfoque Pedagógico:** Teórico-práctico orientado al examen

## Distribución Día por Día

### Día 1: Variables y Tipos de Datos
- **Temas:** Variables primitivas, asignación de memoria y conversiones.
- **Actividades:** Revisión conceptual rápida y resolución de 10 ejercicios de tipado y buenas prácticas.
- **Tiempo Total:** 2 horas.

### Día 2: Control de Flujo: Condicionales
- **Temas:** Estructuras if, else-if, switch y operadores lógicos.
- **Actividades:** Escritura de programas con bifurcaciones lógicas y manejo de casos borde.
- **Tiempo Total:** 2 horas.

### Día 3: Estructuras Repetitivas: Bucles
- **Temas:** Bucles for, while y control de iteraciones (break/continue).
- **Actividades:** Algoritmos de acumulación, contadores y anidamiento básico.
- **Tiempo Total:** 2 horas.

### Día 4: Modularidad: Funciones
- **Temas:** Definición de funciones, paso de parámetros por valor/referencia y retorno de valores.
- **Actividades:** Refactorización de código previo en funciones reutilizables.
- **Tiempo Total:** 2 horas.

### Día 5: Estructuras de Datos: Arrays
- **Temas:** Creación de arreglos, indexación y métodos comunes de recorrido.
- **Actividades:** Algoritmos clásicos de búsqueda lineal y manipulación de listas.
- **Tiempo Total:** 2 horas.

### Día 6: Abstracción: Objetos y Estructuras
- **Temas:** Concepto de clave-valor, diccionarios/objetos literales y métodos asociados.
- **Actividades:** Modelado de entidades del mundo real usando estructuras de datos compuestas.
- **Tiempo Total:** 2 horas.

### Día 7: Programación Orientada a Objetos (POO) y Cierre
- **Temas:** Clases, constructores, atributos, métodos y noción de herencia.
- **Actividades:** Creación de un mini-proyecto integrador y repaso general de dudas.
- **Tiempo Total:** 2 horas.

## Consejos de Preparación
1. Mantén sesiones consistentes sin saltear días.
2. Escribe el código tú mismo sin copiar y pegar soluciones.
3. Dedica los últimos minutos de cada día a anotar puntos no comprendidos."""

    elif version == "V3":
        return """# 1. Resumen de la Situación Académica
- **Materia analizada:** Programación
- **Volumen temático:** 7 ejes fundamentales (Variables, Condicionales, Bucles, Funciones, Arrays, Objetos, POO).
- **Presupuesto temporal:** 7 días continuos × 2 horas/día = **14 horas totales (840 minutos)** disponibles.
- **Diagnóstico del perfil:** Al encontrarse en nivel **Intermedio**, el estudiante no requiere explicaciones introductorias elementales sobre qué es un programa o cómo instalar un editor. El tiempo se optimiza descartando teoría redundante y enfocándose en solidez algorítmica, interacción entre estructuras y encapsulamiento.
- **Supuestos explícitos adoptados:** Se asume el uso de un lenguaje estándar de tipado estructurado u orientado a objetos (ej. Python, JavaScript, Java o C#) y que el entorno de desarrollo ya se encuentra operativo.

---

# 2. Priorización de Contenidos

| Nivel de Prioridad | Ejes Temáticos | Justificación Pedagógica y Complejidad |
|:---|:---|:---|
| **Alta (Núcleo Complejo)** | Funciones (Scope), Arrays (Recorridos) y POO (Clases/Métodos) | Representan el núcleo de la evaluación y el mayor índice de error en exámenes técnicos. |
| **Media (Control de Flujo)** | Condicionales y Bucles Anidados | Requieren dominio de trazas lógicas y algoritmos combinados. |
| **Consolidación (Base)** | Variables y Objetos Literales | Conceptos de rápida asimilación en nivel intermedio; se integran como herramientas de apoyo. |

---

# 3. Plan Organizado por Día (Cronograma Completo del Día 1 al 7)

### Día 1: Variables y Tipos de Datos en Profundidad
* **Objetivo de la sesión:** Dominar alcance de variables (*scope*), mutabilidad, tipado y conversiones de tipo sin errores en tiempo de ejecución.
* **Desglose temporal minuto a minuto (Total: 120 min):**
  - `00:00 - 00:20 (20 min)`: Repaso diagnóstico de tipos primitivos vs. referencia.
  - `00:20 - 00:55 (35 min)`: Práctica de asignación y *casting* de tipos.
  - `00:55 - 01:05 (10 min)`: Pausa activa libre de pantallas.
  - `01:05 - 01:45 (40 min)`: Resolución de 8 ejercicios de lógica con variables y operadores bitwise/aritméticos.
  - `01:45 - 02:00 (15 min)`: Autoevaluación rápida y registro de dudas en cuaderno de notas.

### Día 2: Control de Flujo Lógico (Condicionales y Bucles)
* **Objetivo de la sesión:** Construir algoritmos con bifurcaciones complejas y estructuras repetitivas optimizadas.
* **Desglose temporal minuto a minuto (Total: 120 min):**
  - `00:00 - 00:15 (15 min)`: **Repaso Espaciado Día 1:** Identificar errores de tipado en 3 fragmentos de código.
  - `00:15 - 00:45 (30 min)`: Bloque conceptual de condicionales anidados y bucles (`for`, `while`).
  - `00:45 - 00:55 (10 min)`: Pausa activa cognitiva.
  - `00:55 - 01:45 (50 min)`: Práctica deliberada: 5 desafíos de acumulación, banderas (*flags*) y series numéricas.
  - `01:45 - 02:00 (15 min)`: Autoevaluación: traza manual en papel de un bucle con condición de corte compleja.

### Día 3: Modularización y Funciones
* **Objetivo de la sesión:** Diseñar funciones puras, parametrizadas y reutilizables con control estricto de valores de retorno.
* **Desglose temporal minuto a minuto (Total: 120 min):**
  - `00:00 - 00:15 (15 min)`: **Repaso Espaciado Días 1 y 2:** Refactorizar un bucle simple dentro de un bloque condicional.
  - `00:15 - 00:45 (30 min)`: Firma de funciones, argumentos por defecto, paso por valor vs. referencia y *stack de ejecución*.
  - `00:45 - 00:55 (10 min)`: Pausa activa.
  - `00:55 - 01:45 (50 min)`: Práctica intensiva: descomponer un problema monolítico en 4 funciones coordinadas.
  - `01:45 - 02:00 (15 min)`: Autoevaluación tipo quiz: predicción de salidas (*dry run*).

### Día 4: Estructuras Lineales: Arrays y Algoritmos de Búsqueda
* **Objetivo de la sesión:** Manipular colecciones ordenadas mediante índices, iteraciones y métodos de filtrado y mapeo.
* **Desglose temporal minuto a minuto (Total: 120 min):**
  - `00:00 - 00:15 (15 min)`: **Repaso Espaciado Día 3:** Escribir una función en menos de 5 minutos que calcule promedios.
  - `00:15 - 00:45 (30 min)`: Métodos esenciales de arreglos (inserción, eliminación, recorrido, búsqueda).
  - `00:45 - 00:55 (10 min)`: Pausa activa.
  - `00:55 - 01:45 (50 min)`: Práctica guiada: implementar algoritmo de búsqueda lineal y filtrado de elementos duplicados.
  - `01:45 - 02:00 (15 min)`: Autoevaluación: resolver ejercicio contrarreloj de transformación de arreglos.

### Día 5: Modelado de Datos con Objetos
* **Objetivo de la sesión:** Representar entidades con atributos estructurados y colecciones compuestas de objetos.
* **Desglose temporal minuto a minuto (Total: 120 min):**
  - `00:00 - 00:15 (15 min)`: **Repaso Espaciado Acumulativo (Días 2 y 4):** Iterar un array numérico con condicionales.
  - `00:15 - 00:45 (30 min)`: Estructuras clave-valor, objetos anidados y acceso dinámico a propiedades.
  - `00:45 - 00:55 (10 min)`: Pausa activa.
  - `00:55 - 01:45 (50 min)`: Práctica deliberada: crear un catálogo de productos con búsqueda por identificador y filtro por categoría.
  - `01:45 - 02:00 (15 min)`: Autoevaluación: modelar entidad con propiedades computadas.

### Día 6: Programación Orientada a Objetos (POO)
* **Objetivo de la sesión:** Integrar clases, instancias, constructores y métodos encapsulando estados y comportamientos.
* **Desglose temporal minuto a minuto (Total: 120 min):**
  - `00:00 - 00:15 (15 min)`: **Repaso Espaciado Día 5:** Diferenciar objeto literal vs. instancia de clase.
  - `00:15 - 00:50 (35 min)`: Principios de POO: abstracción, encapsulamiento, métodos getters/setters y relaciones básicas.
  - `00:50 - 01:00 (10 min)`: Pausa activa.
  - `01:00 - 01:45 (45 min)`: Práctica deliberada: modelar sistema de Cuenta Bancaria con validación en métodos de retiro/depósito.
  - `01:45 - 02:00 (15 min)`: Autoevaluación: explicar en voz alta (*Técnica Feynman*) el concepto de constructor.

### Día 7: Simulacro de Examen, Debugging y Repaso Espaciado Integral
* **Objetivo de la sesión:** Validar competencias operativas completas bajo condiciones de tiempo real y consolidar puntos débiles.
* **Desglose temporal minuto a minuto (Total: 120 min):**
  - `00:00 - 00:20 (20 min)`: Repaso mnemotécnico general de sintaxis y conceptos clave (Flashcards).
  - `00:20 - 01:20 (60 min)`: **Simulacro de Examen Completo:** 3 ejercicios prácticos integradores (1 bucle condicional, 1 manipulación de arrays en funciones, 1 clase POO).
  - `01:20 - 01:30 (10 min)`: Pausa activa restauradora.
  - `01:30 - 01:50 (20 min)`: Corrección cruzada, detección de cuellos de botella y *debugging*.
  - `01:50 - 02:00 (10 min)`: Conclusiones finales y cierre estratégico.

---

# 4. Desglose y Presupuesto Temporal Estimado por Actividad

| Día | Eje Temático Principal | Repaso Activo (min) | Bloque Conceptual (min) | Pausa Cognitiva (min) | Práctica Deliberada (min) | Autoevaluación (min) | Total Sesión |
|:---:|:---|:---:|:---:|:---:|:---:|:---:|:---:|
| 1 | Variables y Tipos de Datos | 0 | 35 | 10 | 55 | 20 | **120 min (2h)** |
| 2 | Condicionales y Bucles | 15 | 30 | 10 | 50 | 15 | **120 min (2h)** |
| 3 | Funciones y Modularidad | 15 | 30 | 10 | 50 | 15 | **120 min (2h)** |
| 4 | Arrays y Colecciones | 15 | 30 | 10 | 50 | 15 | **120 min (2h)** |
| 5 | Objetos y Diccionarios | 15 | 30 | 10 | 50 | 15 | **120 min (2h)** |
| 6 | Programación Orientada a Objetos | 15 | 35 | 10 | 45 | 15 | **120 min (2h)** |
| 7 | Simulacro de Examen y Consolidación | 20 | 0 | 10 | 70 | 20 | **120 min (2h)** |
| **Total** | **Plan de 7 Jornadas** | **95 min** | **190 min** | **70 min** | **370 min** | **115 min** | **840 min (14h)** |

---

# 5. Objetivos de Aprendizaje de Cada Sesión

1. **Sesión 1:** Explicar el ciclo de vida de una variable en memoria y aplicar conversiones explícitas de tipos.
2. **Sesión 2:** Evaluar expresiones booleanas compuestas e implementar bucles sin caer en iteraciones infinitas.
3. **Sesión 3:** Aislar responsabilidades algorítmicas en funciones puras con manejo correcto de parámetros y retornos.
4. **Sesión 4:** Filtrar, mapear y ordenar colecciones de datos secuenciales aplicando buenas prácticas algorítmicas.
5. **Sesión 5:** Estructurar datos multidimensionales mediante objetos y consultar propiedades anidadas.
6. **Sesión 6:** Diseñar clases con estados internos protegidos y métodos que implementen reglas de negocio.
7. **Sesión 7:** Resolver desafíos de examen bajo presión temporal demostrando capacidad de autodiagnóstico y *debugging*.

---

# 6. Técnicas de Estudio Recomendadas

* **Técnica Pomodoro Adaptada (40/10):** Para ajustarse a la preferencia de sesiones cortas, cada jornada se fracciona en dos bloques focales de 40 a 50 minutos separados por 10 minutos de reposo absoluto de pantallas para facilitar la consolidación sináptica.
* **Active Recall (Evocación Activa):** Antes de consultar apuntes o documentación, intentar escribir el código o explicar la sintaxis desde la memoria.
* **Técnica Feynman:** Tomar conceptos complejos (como la diferencia entre un objeto y una clase) y explicarlos en voz alta con analogías simples como si se enseñara a un principiante.
* **Rubber Duck Debugging:** Al encontrar un fallo en el código, verbalizar línea por línea lo que el programa debería hacer antes de recurrir a ayudas externas.

---

# 7. Momentos y Estrategia de Repaso Espaciado

Se combate la **Curva del Olvido de Ebbinghaus** mediante una regla de tres impactos:
1. **Primer Impacto:** Adquisición y práctica inmediata (el día del tema).
2. **Segundo Impacto (a las 24-48 horas):** Bloque inicial de 15 minutos al inicio del día siguiente o subsiguiente con ejercicios de recuperación activa.
3. **Tercer Impacto (Consolidación pre-examen):** El Día 7 integra todos los temas en un simulacro global, forzando al cerebro a conectar conceptos dispersos.

---

# 8. Autoevaluaciones, Retos y Ejercicios Prácticos

* **Reto Día 2:** Escribir un algoritmo que recorra los números del 1 al 50 e imprima múltiplos de 3 y 5 con condiciones específicas (FizzBuzz adaptado).
* **Reto Día 4:** Crear una función que reciba un array de calificaciones y devuelva un nuevo array únicamente con las notas aprobatorias ordenadas de mayor a menor.
* **Reto Día 6:** Implementar la clase `Estudiante` con atributos `nombre`, `matricula` y lista de `materias`, con un método `agregarCalificacion()` que valide que la nota esté entre 0 y 10.
* **Simulacro Día 7:** 3 problemas cronometrados que combinen arrays de objetos procesados mediante funciones modulares.

---

# 9. Recomendaciones Finales y Gestión del Descanso

* **La regla de las 12 horas previas:** La noche previa al examen no programar sesiones de estudio intensivas; suspender la actividad intelectual pesada para garantizar 8 horas de sueño profundo (etapa donde se consolida la memoria de largo plazo).
* **Estrategia en el examen:** Leer la totalidad de las consignas durante los primeros 5 minutos, resolver primero los problemas de menor complejidad para ganar confianza y modularizar en pseudocódigo antes de escribir código definitivo.
* **Gestión de la frustración:** Si un algoritmo no funciona, aislar los valores de las variables imprimiéndolos en consola en lugar de reescribir todo el bloque a ciegas."""

def ejecutar_prompt(prompt_texto, version_label):
    """
    Ejecuta el prompt consultando a OpenAI si hay credenciales configuradas;
    de lo contrario, recurre a la simulación académica registrada.
    """
    api_key = os.getenv("OPENAI_API_KEY")
    modelo = os.getenv("OPENAI_MODEL", "gpt-4o-mini")
    
    if api_key and not api_key.startswith("sk-proj-xxxx") and len(api_key) > 20:
        try:
            from openai import OpenAI
            cliente = OpenAI(api_key=api_key)
            print(f"[API OpenAI]: Enviando solicitud {version_label} a modelo {modelo}...")
            respuesta = cliente.chat.completions.create(
                model=modelo,
                messages=[{"role": "user", "content": prompt_texto}],
                temperature=0.7
            )
            print(f"[API OpenAI]: Respuesta recibida exitosamente.")
            return respuesta.choices[0].message.content
        except Exception as e:
            print(f"[Aviso de Conexión/API]: {e}")
            print(f"[Fallback]: Activando Modo Simulación Académica Determinista para {version_label}.")
            return simular_respuesta_academica(version_label)
    else:
        print(f"[Modo Determinista]: Entregando resultado experimental registrado para {version_label}.")
        return simular_respuesta_academica(version_label)
````

| **V3 (Optimizado)** | Fast Prompting integral: Few-Shot, Negative Constraints, Task Decomposition y anclaje cognitivo. | Excelente (9 secciones obligatorias, tablas consolidadas, minutos exactos). | Máxima (adaptación a sesiones cortas, nivel intermedio y 2 horas exactas). | Excelente (jerárquica, operativa y didácticamente sólida). | Cumple rigurosamente con los 9 bloques solicitados; es un plan inmediatamente aplicable por el alumno. |
